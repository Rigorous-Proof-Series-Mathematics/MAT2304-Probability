### Definition — Field of Probability

A system of sets$\mathcal{F}$ together with an assignment of numbers $P(A)$ satisfying Axioms I–V is called a field of probability.
<!-- block:end -->

<!-- block:start proposition label=construction-of-finite-probability-fields -->

### Proposition — Construction of Finite Probability Fields

Let

$$
E=\{\xi_1,\xi_2,\dots,\xi_k\}
$$

be a finite set of elementary events.

Let

$$
p_1,p_2,\dots,p_k
$$

be non-negative real numbers satisfying

$$
p_1+p_2+\cdots+p_k=1.
$$

Take

$$
\mathcal{F}
$$

to be the set of all subsets of \(E\).

Define

$$
P\bigl(\{\xi_{i_1},\xi_{i_2},\dots,\xi_{i_n}\}\bigr)
=
p_{i_1}+p_{i_2}+\cdots+p_{i_n}.
$$

Then

$$
(\mathcal{F},P)
$$

forms a finite probability space.

<!-- block:start axiom label=kolmogorov-field-of-events -->
### Field of Events

Let $E$ be a collection of elementary events $ \xi,\eta,\zeta,\dots $ and let $\mathcal{F}$ be a collection of subsets of $E$, called random events.

Then $ \mathcal{F} $ is a field of sets.
<!-- block:end -->

<!-- block:start axiom label=kolmogorov-total-space -->
### Total Space

The field of events contains the total set of elementary events:
$
E \in \mathcal{F}
$
<!-- block:end -->

<!-- block:start axiom label=kolmogorov-probability-measure -->
### Probability Measure

To every set $ A \in \mathcal{F} $ there corresponds a non-negative real number $P(A)$ called the probability of the event $A$.

Moreover, $P(A) \ge 0$
<!-- block:end -->

<!-- block:start axiom label=kolmogorov-normalization -->
### Normalization

The probability of the total space equals one: 
$$P(E)=1$$
<!-- block:end -->

<!-- block:start axiom label=kolmogorov-additivity -->
### Finite Additivity

If two events $A,B \in \mathcal{F}$ have no elements in common,$A \cap B = \varnothing$ then

$$
P(A+B)=P(A)+P(B)
$$
<!-- block:end -->

<!-- block:start remark label=consistency-of-axioms -->
### Consistency of the Axioms

The system of axioms is consistent.

For example, let $E=\{\xi\}$ and let $\mathcal{F}=\{E,\varnothing\}$ with 
$$P(E)=1,\qquad
P(\varnothing)=0
$$

Then all axioms are satisfied.
<!-- block:end -->

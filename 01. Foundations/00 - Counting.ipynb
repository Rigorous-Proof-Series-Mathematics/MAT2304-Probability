### Combinations Without Repetition

The number of unordered selections of $k$ objects from $n$ distinct objects is

$$
\binom{n}{k}
=
\frac{n!}{k!(n-k)!}.
$$

<!-- proof:start -->

### Proof

Begin with the number of ordered selections:

$$
\frac{n!}{(n-k)!}.
$$

Each unordered selection corresponds to exactly $k!$ reorderings.

Therefore:

$$
\binom{n}{k}
=
\frac{n!}{k!(n-k)!}.
$$

<!-- block:end -->

<!-- block:start definition label=combinations-with-repetition -->

### Definition — Combinations With Repetition

The number of unordered selections of $k$ objects from $n$ types when repetition is allowed is

$$
\binom{n+k-1}{k}.
$$

<!-- proof:start -->

### Proof

Represent the $k$ chosen objects by stars and separate the $n$ categories using $n-1$ bars.

The total number of symbols is

$$
k+n-1.
$$

Choosing the positions of the $k$ stars gives

$$
\binom{n+k-1}{k}.
$$

<!-- block:end -->

<!-- block:start definition label=permutations -->

### Permutations of Distinct Objects

The number of ways to arrange $n$ distinct objects in order is

$$
n!.
$$

<!-- proof:start -->

### Proof

There are:

- $n$ choices for the first position,
- $n-1$ for the second,
- continuing down to $1$.

Thus:

$$
n(n-1)(n-2)\cdots1=n!.
$$

<!-- block:end -->

<!-- block:start definition label=partial-permutations -->

### Partial Permutations

The number of ordered selections of $k$ objects chosen from $n$ distinct objects is

$$
P(n,k)
=
\frac{n!}{(n-k)!}.
$$

<!-- proof:start -->

### Proof

There are:

$$
n
$$

choices for the first position,

$$
n-1
$$

for the second,

continuing until

$$
n-k+1.
$$

Hence:

$$
n(n-1)\cdots(n-k+1)
=
\frac{n!}{(n-k)!}.
$$

<!-- block:end -->

<!-- block:start definition label=permutations-with-repetition -->

### Permutations With Repetition

Suppose among $n$ objects there are

$$
n_1,n_2,\dots,n_r
$$

indistinguishable copies of each type, with

$$
n_1+\cdots+n_r=n.
$$

The number of distinct permutations is

$$
\frac{n!}{n_1!n_2!\cdots n_r!}.
$$

<!-- proof:start -->

### Proof

If all objects were distinct there would be

$$
n!
$$

permutations.

However, permuting identical objects among themselves does not create new arrangements.

For each type $i$, there are

$$
n_i!
$$

redundant reorderings.

Dividing removes the overcounting.

<!-- block:end -->

<!-- block:start definition label=stars-and-bars -->

### Stars and Bars

The number of nonnegative integer solutions of

$$
x_1+x_2+\cdots+x_n=k
$$

is

$$
\binom{n+k-1}{k}.
$$

<!-- proof:start -->

### Proof

Represent the $k$ units by stars and separate the $n$ variables using $n-1$ bars.

Each arrangement uniquely determines a solution.

The total number of symbols is

$$
k+n-1.
$$

Choosing the positions of the stars gives

$$
\binom{n+k-1}{k}.
$$

<!-- block:end -->

<!-- block:start definition label=rule-of-product -->

### Rule of Product

If a procedure consists of successive independent choices with

$$
n_1,n_2,\dots,n_k
$$

possible outcomes respectively, then the total number of outcomes is

$$
n_1n_2\cdots n_k.
$$

<!-- proof:start -->

### Proof

For each of the $n_1$ possibilities of the first choice, there are $n_2$ possibilities for the second choice, and so on.

Multiplying recursively gives

$$
n_1n_2\cdots n_k.
$$

<!-- block:end -->

<!-- block:start definition label=rule-of-sum -->

### Rule of Sum

If a task can occur in one of several mutually exclusive ways with counts

$$
n_1,n_2,\dots,n_k,
$$

then the total number of possibilities is

$$
n_1+n_2+\cdots+n_k.
$$

<!-- proof:start -->

### Proof

Since the possibilities are mutually exclusive, every valid outcome belongs to exactly one class.

Therefore the total number is the sum of the cardinalities.

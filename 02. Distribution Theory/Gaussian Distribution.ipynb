### Definition — Gaussian Density

A random variable $X$ is said to follow a Gaussian distribution with mean $\mu$ and variance $\sigma^2>0$ if its probability density function is

$$
f(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right).
$$

<!-- proof:start -->

### Proof

The normalization constant is chosen so that

$$
\int_{-\infty}^{\infty} f(x)\,dx=1.
$$

Substituting

$$
z=\frac{x-\mu}{\sigma}
$$

reduces the integral to the standard Gaussian integral

$$
\int_{-\infty}^{\infty} e^{-z^2/2}\,dz=\sqrt{2\pi}.
$$

<!-- block:end -->

<!-- block:start theorem label=gaussian-integral -->

### Theorem — Gaussian Integral

The Gaussian integral satisfies

$$
\int_{-\infty}^{\infty} e^{-x^2}\,dx
=
\sqrt{\pi}.
$$

<!-- proof:start -->

### Proof

Let

$$
I=\int_{-\infty}^{\infty} e^{-x^2}\,dx.
$$

Then

$$
I^2
=
\int_{-\infty}^{\infty}
\int_{-\infty}^{\infty}
e^{-(x^2+y^2)}\,dx\,dy.
$$

Passing to polar coordinates:

$$
x=r\cos\theta,
\qquad
y=r\sin\theta,
$$

with Jacobian $r$:

$$
I^2
=
\int_0^{2\pi}
\int_0^\infty
e^{-r^2}r\,dr\,d\theta.
$$

Evaluating:

$$
I^2
=
2\pi\cdot\frac12
=
\pi.
$$

Therefore:

$$
I=\sqrt{\pi}.
$$

<!-- block:start proposition label=gaussian-mean -->

### Proposition — Mean of a Gaussian Distribution

If

$$
X\sim\mathcal N(\mu,\sigma^2),
$$

then

$$
\mathbb E[X]=\mu.
$$

<!-- proof:start -->

### Proof

By definition:

$$
\mathbb E[X]
=
\int_{-\infty}^{\infty}
x f(x)\,dx.
$$

Substitute

$$
z=\frac{x-\mu}{\sigma},
\qquad
x=\mu+\sigma z.
$$

Then:

$$
\mathbb E[X]
=
\mu
+
\sigma
\int_{-\infty}^{\infty}
z
\frac{e^{-z^2/2}}{\sqrt{2\pi}}dz.
$$

The second integral vanishes because the integrand is odd.

Hence:

$$
\mathbb E[X]=\mu.
$$

<!-- block:end -->

<!-- block:start proposition label=gaussian-variance -->

### Proposition — Variance of a Gaussian Distribution

If

$$
X\sim\mathcal N(\mu,\sigma^2),
$$

then

$$
\mathrm{Var}(X)=\sigma^2.
$$

<!-- proof:start -->

### Proof

Compute:

$$
\mathrm{Var}(X)
=
\mathbb E[(X-\mu)^2].
$$

Using the substitution

$$
z=\frac{x-\mu}{\sigma},
$$

we obtain

$$
\mathrm{Var}(X)
=
\sigma^2
\int_{-\infty}^{\infty}
z^2
\frac{e^{-z^2/2}}{\sqrt{2\pi}}dz.
$$

The remaining integral equals $1$, yielding

$$
\mathrm{Var}(X)=\sigma^2.
$$

<!-- block:end -->

<!-- block:start theorem label=standardization -->

### Theorem — Standardization of a Gaussian Variable

If $ X\sim\mathcal N(\mu,\sigma^2)$, then $ Z=\frac{X-\mu}{\sigma}$ satisfies $Z\sim\mathcal N(0,1)$.

<!-- proof:start -->

### Proof

Using the change-of-variables formula:

$$
f_Z(z)
=
f_X(\mu+\sigma z)\left|\frac{dx}{dz}\right|.
$$

Since

$$
\frac{dx}{dz}=\sigma,
$$

we obtain

$$
f_Z(z)
=
\frac1{\sqrt{2\pi}}
e^{-z^2/2}.
$$

Thus $Z$ follows the standard normal distribution.

<!-- block:end -->

<!-- block:start theorem label=sum-of-independent-gaussians -->

### Theorem — Sum of Independent Gaussian Variables

If

$$
X\sim\mathcal N(\mu_1,\sigma_1^2),
\qquad
Y\sim\mathcal N(\mu_2,\sigma_2^2),
$$

and $X$ and $Y$ are independent, then

$$
X+Y
\sim
\mathcal N(\mu_1+\mu_2,\sigma_1^2+\sigma_2^2).
$$

<!-- proof:start -->

### Proof

The characteristic function of a Gaussian variable is

$$
\phi_X(t)
=
\exp\left(it\mu_1-\frac12\sigma_1^2 t^2\right).
$$

Similarly for $Y$.

Independence implies

$$
\phi_{X+Y}(t)
=
\phi_X(t)\phi_Y(t).
$$

Multiplying gives

$$
\phi_{X+Y}(t)
=
\exp\left(
it(\mu_1+\mu_2)
-
\frac12(\sigma_1^2+\sigma_2^2)t^2
\right).
$$

This is the characteristic function of a Gaussian distribution.

<!-- block:end -->

<!-- block:start theorem label=multivariate-gaussian-independent -->

### Theorem — Multivariate Gaussian Distribution of Independent Variables

Let

$$
X_1,X_2,\dots,X_n
$$

be independent Gaussian random variables with

$$
X_i\sim\mathcal N(\mu_i,\sigma_i^2).
$$

Then the random vector

$$
\mathbf X=
\begin{bmatrix}
X_1\\
X_2\\
\vdots\\
X_n
\end{bmatrix}
$$

follows a multivariate Gaussian distribution with mean vector

$$
\mu=
\begin{bmatrix}
\mu_1\\
\mu_2\\
\vdots\\
\mu_n
\end{bmatrix}
$$

and covariance matrix

$$
\Sigma=
\begin{bmatrix}
\sigma_1^2 &0&\cdots&0\\
0&\sigma_2^2&\cdots&0\\
\vdots&\vdots&\ddots&\vdots\\
0&0&\cdots&\sigma_n^2
\end{bmatrix}.
$$

Its density is

$$
p(\mathbf x)
=
\frac1{(2\pi)^{n/2}|\Sigma|^{1/2}}
\exp\left(
-\frac12
(\mathbf x-\mu)^T
\Sigma^{-1}
(\mathbf x-\mu)
\right).
$$

<!-- proof:start -->

### Proof

Independence implies

$$
p(\mathbf x)
=
\prod_{i=1}^n
\frac1{\sqrt{2\pi\sigma_i^2}}
\exp\left(
-\frac{(x_i-\mu_i)^2}{2\sigma_i^2}
\right).
$$

Combining the exponentials gives

$$
p(\mathbf x)
=
\frac1{(2\pi)^{n/2}\prod_i\sigma_i}
\exp\left(
-\frac12
\sum_i
\frac{(x_i-\mu_i)^2}{\sigma_i^2}
\right).
$$

Recognizing

$$
|\Sigma|^{1/2}
=
\prod_i\sigma_i
$$

and

$$
(\mathbf x-\mu)^T\Sigma^{-1}(\mathbf x-\mu)
=
\sum_i
\frac{(x_i-\mu_i)^2}{\sigma_i^2},
$$

we obtain the multivariate Gaussian form.

<!-- block:end -->

<!-- block:start theorem label=multivariate-gaussian-correlated -->

### Theorem — Multivariate Gaussian Distribution of Correlated Variables

Let

$$
\mathbf X=
\begin{bmatrix}
X_1\\
X_2\\
\vdots\\
X_n
\end{bmatrix}
$$

be a Gaussian random vector with mean vector

$$
\mu
$$

and covariance matrix

$$
\Sigma.
$$

Then its density is

$$
p(\mathbf x)
=
\frac1{(2\pi)^{n/2}|\Sigma|^{1/2}}
\exp\left(
-\frac12
(\mathbf x-\mu)^T
\Sigma^{-1}
(\mathbf x-\mu)
\right).
$$

<!-- proof:start -->

### Proof

Unlike the independent case, the covariance matrix contains nonzero off-diagonal terms:

$$
\Sigma_{ij}
=
\mathrm{Cov}(X_i,X_j).
$$

Diagonalizing the covariance matrix:

$$
\Sigma
=
Q\Lambda Q^T,
$$

and defining

$$
\mathbf y
=
Q^T(\mathbf x-\mu),
$$

the quadratic form becomes

$$
(\mathbf x-\mu)^T\Sigma^{-1}(\mathbf x-\mu)
=
\mathbf y^T\Lambda^{-1}\mathbf y.
$$

The transformed variables become independent Gaussian modes.

The determinant factor arises from the Jacobian of the transformation.

<!-- block:end -->

### Proposition — Marginal Distribution in the Independent Gaussian Case

Let

$$
\mathbf X=
(X_1,\dots,X_n)
$$

be a multivariate Gaussian vector with diagonal covariance matrix.

Then the marginal distribution of any component \(X_i\) is

$$
X_i\sim\mathcal N(\mu_i,\sigma_i^2).
$$

<!-- proof:start -->

### Proof

Because the variables are independent, the joint density factorizes:

$$
p(\mathbf x)
=
\prod_{j=1}^n p_j(x_j).
$$

The marginal density of \(X_i\) is obtained by integrating over all remaining variables:

$$
p_i(x_i)
=
\int
p(\mathbf x)
\prod_{j\neq i}dx_j.
$$

Since all factors integrate to \(1\),

$$
p_i(x_i)
=
\frac1{\sqrt{2\pi\sigma_i^2}}
\exp\left(
-\frac{(x_i-\mu_i)^2}{2\sigma_i^2}
\right).
$$

<!-- block:end -->

<!-- block:start theorem label=marginal-correlated-gaussian -->

### Theorem — Marginal Distribution of a Correlated Multivariate Gaussian

Let

$$
\mathbf X=
\begin{bmatrix}
\mathbf X_a\\
\mathbf X_b
\end{bmatrix}
$$

be jointly Gaussian with mean

$$
\mu=
\begin{bmatrix}
\mu_a\\
\mu_b
\end{bmatrix}
$$

and covariance matrix

$$
\Sigma=
\begin{bmatrix}
\Sigma_{aa} & \Sigma_{ab}\\
\Sigma_{ba} & \Sigma_{bb}
\end{bmatrix}.
$$

Then the marginal distribution of \(\mathbf X_a\) is Gaussian:

$$
\mathbf X_a
\sim
\mathcal N(\mu_a,\Sigma_{aa}).
$$

<!-- proof:start -->

### Proof

The marginal density is obtained by integrating the joint density over the variables in \(\mathbf X_b\):

$$
p(\mathbf x_a)
=
\int
p(\mathbf x_a,\mathbf x_b)\,
d\mathbf x_b.
$$

Since Gaussian integrals are closed under integration, the result remains Gaussian.

The resulting mean vector is the corresponding subvector:

$$
\mu_a,
$$

and the covariance becomes the corresponding covariance block:

$$
\Sigma_{aa}.
$$

Thus marginals of multivariate Gaussians are themselves Gaussian.

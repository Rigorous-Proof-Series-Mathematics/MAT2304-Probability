<!-- block:start axiom label=Field of Events -->
Let $begin:math:text$E$end:math:text$ be a collection of elementary events

$$
\xi,\eta,\zeta,\dots
$$

and let $begin:math:text$\\mathcal\{F\}$end:math:text$ be a collection of subsets of $begin:math:text$E$end:math:text$,
called random events.

Then

$$
\mathcal{F}
$$

is a field of sets.

### Axiom I — Field of Events

Let $E$ be a collection of elementary events

$$
\xi,\eta,\zeta,\dots
$$

and let $\mathcal{F}$ be a collection of subsets of $E$,
called random events.

Then

$$
\mathcal{F}
$$

is a field of sets.
<!-- block:end -->

<!-- block:start axiom label=kolmogorov-total-space -->
### Axiom II — Total Space

The field of events contains the total set of elementary events:

$$
E \in \mathcal{F}
$$
<!-- block:end -->

<!-- block:start axiom label=kolmogorov-probability-measure -->
### Axiom III — Probability Measure

To every set

$$
A \in \mathcal{F}
$$

there corresponds a non-negative real number

$$
P(A)
$$

called the probability of the event $begin:math:text$A$end:math:text$.

Moreover,

$$
P(A) \ge 0
$$
